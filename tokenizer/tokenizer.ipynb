```python
import tiktoken

encoding = tiktoken.encoding_for_model("gpt-4.1-mini")
my_str = "Hi my name is Viswanath and I like to learn new things around Generatove AI"
tokens = encoding.encode(my_str)
```

```python
print("Number of words ", len(my_str.split()))
print("Number of tokens ", len(tokens))
```

```python
for token_id in tokens:
    token_text = encoding.decode([token_id])
    print(f"{token_id} = '{token_text}'")
```
